```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import kagglehub
import os
import re
import nltk
from nltk.corpus import stopwords

# --- Start of code to ensure jobs_df is defined and processed ---
# Re-download the jobs dataset to get its correct path (will use cached version)
jobs_data_dir = kagglehub.dataset_download("kshitizregmi/jobs-and-job-description")
jobs_csv_file_path = os.path.join(jobs_data_dir, 'job_title_des.csv')
jobs_df = pd.read_csv(jobs_csv_file_path)

# Ensure NLTK data is downloaded for stopwords
try:
    nltk.data.find('corpora/stopwords')
except LookupError:
    nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

def clean_text(text):
    text = str(text).lower()
    text = re.sub(r'http\S+|www\S+', '', text)
    text = re.sub(r'\S+@\S+', '', text)
    text = re.sub(r'[^a-z0-9\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

def remove_stopwords(text):
    if not isinstance(text, str): return ""
    return ' '.join([word for word in text.split() if word not in stop_words])

# Apply preprocessing steps to jobs_df
jobs_df['cleaned_job_description'] = jobs_df['Job Description'].apply(clean_text)
jobs_df['lemmatized_job_description'] = jobs_df['cleaned_job_description'].apply(remove_stopwords) # Renamed to 'lemmatized_job_description' but only contains cleaned and stopword-removed text
# --- End of code to ensure jobs_df is defined and processed ---


# 1. Ensure no NaN values in your cleaned column
jobs_df['lemmatized_job_description'] = jobs_df['lemmatized_job_description'].fillna('')

# 2. Initialize and fit the TF-IDF Vectorizer
# - max_features=5000 limits vocabulary to the top 5,000 important words
# - ngram_range=(1, 2) includes both unigrams ("flutter") and bigrams ("flutter developer")
tfidf = TfidfVectorizer(max_features=5000, ngram_range=(1, 2))
tfidf_job_matrix = tfidf.fit_transform(jobs_df['lemmatized_job_description'])

print(f"TF-IDF Matrix Shape: {tfidf_job_matrix.shape}")

# 3. Recommendation function
def recommend_jobs(cleaned_resume_text, tfidf_model, job_matrix, df, top_n=5):
    """
    Transforms a preprocessed resume string and returns the top N matching jobs
    ranked by cosine similarity score.
    """
    # Vectorize the incoming resume text using the fitted vocabulary
    resume_vector = tfidf_model.transform([cleaned_resume_text])

    # Compute Cosine Similarity between the resume and all jobs
    scores = cosine_similarity(resume_vector, job_matrix).flatten()

    # Get top N indices sorted in descending order
    top_indices = scores.argsort()[::-1][:top_n]

    # Retrieve matching rows and attach similarity score
    recommendations = df.iloc[top_indices].copy()
    recommendations['similarity_score'] = scores[top_indices]

    return recommendations
```

```python
sample_resume_text = "looking for python django backend developer role with sql database experience"

# 2. Get top 5 recommendations
results = recommend_jobs(sample_resume_text, tfidf, tfidf_job_matrix, jobs_df, top_n=5)

# 3. Format similarity score as percentage
results['match_percentage'] = (results['similarity_score'] * 100).round(2).astype(str) + '%'

# 4. Display the top results clearly
print("--- TOP RECOMMENDED JOBS (TF-IDF - Updated Preprocessing) ---")
for idx, row in results.reset_index(drop=True).iterrows():
    print(f"\nMatch #{idx+1} | Score: {row['match_percentage']}")
    print(f"Job Description Snippet: {row['lemmatized_job_description'][:200]}...")
```

```python
import numpy as np

# Apply relative scaling so the top score feels like a "percentage fit"
def scale_match_score(raw_score, max_expected=0.5):
    # Clips max expected cosine score (e.g. 0.5) to 100%
    scaled = (raw_score / max_expected) * 100
    return np.clip(scaled, 0, 100).round(2)
```

```python
import pandas as pd
import numpy as np
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

# 1. Load the lightweight, high-performance semantic model
model = SentenceTransformer('all-MiniLM-L6-v2')

# 2. Ensure text column has no NaN values
jobs_df['lemmatized_job_description'] = jobs_df['lemmatized_job_description'].fillna('')

# 3. Generate dense vector embeddings for all job descriptions
# (Takes 10–30 seconds depending on dataset size)
print("Generating job embeddings with updated preprocessing...")
job_embeddings = model.encode(jobs_df['lemmatized_job_description'].tolist(), show_progress_bar=True)

# 4. Semantic Recommendation Function
def recommend_jobs_semantic(resume_text, embedding_model, job_embeddings, df, top_n=5):
    # Encode the input resume text into a semantic vector
    resume_embedding = embedding_model.encode([resume_text])

    # Calculate Cosine Similarity
    scores = cosine_similarity(resume_embedding, job_embeddings).flatten()

    # Get top N matches
    top_indices = scores.argsort()[::-1][:top_n]

    # Format results
    recommendations = df.iloc[top_indices].copy()
    recommendations['similarity_score'] = scores[top_indices]
    recommendations['match_percentage'] = (recommendations['similarity_score'] * 100).round(2).astype(str) + '%'

    return recommendations[['match_percentage', 'similarity_score', 'lemmatized_job_description']]

# 5. Test with your sample resume
sample_resume_text = "looking for python django backend developer role with sql database experience"
matches = recommend_jobs_semantic(sample_resume_text, model, job_embeddings, jobs_df, top_n=5)

# Display results
matches
```

```python
sample_resume_data_scientist = "seeking a data scientist role with strong experience in machine learning, deep learning, Python, and SQL, preferably in a research and development environment."

# Get top 5 recommendations using the semantic model
matches_data_scientist = recommend_jobs_semantic(sample_resume_data_scientist, model, job_embeddings, jobs_df, top_n=5)

# Apply the scaling function to the 'similarity_score'
# Assuming a max_expected cosine similarity of 0.6 for semantic models often yields reasonable scaling
max_expected_semantic_score = 0.6
matches_data_scientist['scaled_match_percentage'] = matches_data_scientist['similarity_score'].apply(lambda x: scale_match_score(x, max_expected=max_expected_semantic_score))

# Display the results with the new scaled percentage
print("--- TOP RECOMMENDED JOBS for Data Scientist Resume (Semantic - Updated Preprocessing) ---")
for idx, row in matches_data_scientist.reset_index(drop=True).iterrows():
    print(f"\nMatch #{idx+1} | Scaled Score: {row['scaled_match_percentage']:.2f}% | Raw Score: {row['similarity_score']:.4f}")
    print(f"Job Description Snippet: {row['lemmatized_job_description'][:200]}...")
```
