```python
students = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Alice', 'Bob', 'Charlie']
})

scores = pd.DataFrame({
    'ID': [1, 2, 3],
    'Score': [85, 90, 78]
})

courses = pd.DataFrame({
    'ID': [1, 2, 3],
    'Course': ['Math', 'Science', 'History']
})
merged_df = students.merge(scores, on='ID').merge(courses, on='ID')

print(merged_df)
```

```python
import numpy as np
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [24, np.nan, 22, 28],
    'Score': [85, 90, np.nan, 75]
}
df = pd.DataFrame(data)
print("Original:\n", df)
df_filled = df.copy()
df_filled['Age'] = df_filled['Age'].fillna(df_filled['Age'].mean())
df_filled['Score'] = df_filled['Score'].fillna(df_filled['Score'].median())
print("\nFilled:\n", df_filled)
df_dropped_rows = df.dropna()          # rows with any NaN removed
df_dropped_cols = df.dropna(axis=1)    # columns with any NaN removed

print("\nDropped rows:\n", df_dropped_rows)
print("\nDropped columns:\n", df_dropped_cols)
```

```python
data = {
    'Employee': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Department': ['Sales', 'IT', 'HR', 'IT', 'Sales'],
    'Salary': [55000, 72000, 63000, 48000, 80000],
    'Years_of_Service': [3, 6, 7, 4, 9]
}
df = pd.DataFrame(data)

filtered_df = df[(df['Salary'] > 60000) & (df['Years_of_Service'] > 5)]
print(filtered_df)
result = df.loc[df['Salary'] > 60000, ['Employee', 'Salary']]
print(result)
```

```python

data = {
    'Employee_Name': ['Alice Johnson', 'Bob Smith', 'Charlie Brown', 'David Miller']
}
df = pd.DataFrame(data)

df[['First_Name', 'Last_Name']] = df['Employee_Name'].str.split(' ', expand=True)

df['Initials'] = df['First_Name'].str[0] + df['Last_Name'].str[0]

print(df)
```

```python

data = {
    'Student': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Grade': [85, 45, 62, 30, 78]
}
df = pd.DataFrame(data)

df['Pass/Fail'] = df['Grade'].apply(lambda x: 'Pass' if x >= 50 else 'Fail')

def categorize(grade):
    if grade >= 90:
        return 'A'
    elif grade >= 75:
        return 'B'
    elif grade >= 60:
        return 'C'
    elif grade >= 50:
        return 'D'
    else:
        return 'F'

df['Letter_Grade'] = df['Grade'].apply(categorize)

print(df)
```

```python

data = {
    'Product': ['Pen', 'Notebook', 'Bag', 'Bottle', 'Pencil', 'Folder', 'Marker'],
    'Category': ['Stationery', 'Stationery', 'Accessories', 'Accessories', 'Stationery', 'Office', 'Stationery']
}
df = pd.DataFrame(data)
df['Category'] = df['Category'].astype('category')
freq = df['Category'].value_counts()
print(freq)
print("\nDtype check:", df['Category'].dtype)

freq.plot(kind='bar', color='teal')
plt.title('Category Frequency')
plt.xlabel('Category')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python

data = {
    'Product': ['Pen', 'Notebook', 'Pen', 'Bag', 'Notebook', 'Bag'],
    'Sales': [200, 450, 300, 700, 350, 600],
    'Quantity': [20, 15, 30, 10, 12, 8],
    'Cost': [120, 300, 180, 400, 250, 350]
}
df = pd.DataFrame(data)
def total_profit(group):
    return (group['Sales'] - group['Cost']).sum()

profit_df = df.groupby('Product').apply(total_profit, include_groups=False).reset_index(name='Profit')

print(profit_df)
```
