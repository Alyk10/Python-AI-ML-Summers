```python
x = np.array([0, 1, 2, 3, 4, 5])
y = np.array([0, 10, 20, 30, 40, 50])

plt.plot(x, y)
plt.grid(True) # This line adds the grid
plt.title('Plot with Grid')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```
